##reading file hostel_bois.txt:


```python
f=open('hostel_bois.txt','r',encoding='utf-8')
```

```python
#reading the file:
content=f.read()  #"using this read() the pionter goes to end"
print(content)
```

##spliting into parse:


```python
lines=content.split('\n')
parsed_messages=[] # list to store data as dictionary
for line in lines:
  timestamp,rest=line.split(" - ",1)
  if ": " in rest:
    sender,message=rest.split(": ",1)
    #strip message for remove \n .
    message=message.strip()
    # append each information in dictionary formate.
    parsed_messages.append({
        'timestamp':timestamp,
        'sender':sender,
        'message':message
    })
```

```python
print(parsed_messages)
```

```python
#calculating how many message :
total_message=len(parsed_messages)
print(total_message)
```

```python
#how many user :
users=set ()

for msg in parsed_messages:
  users.add(msg['sender'])

print(users)
```

```python
# per sender message count:
count={}

for msg in parsed_messages:
  sender=msg['sender']
  if sender in count:
    # if present then increase count or nor then simple add and assign 1
    count[sender]+=1
  else:
    count[sender]=1

# sort data
sorted_count=sorted(count.items(), key=lambda x: x[1],reverse=True)
print(sorted_count)
```

```python
# counting days :
from datetime import datetime
# first date
first_date=datetime.strptime(parsed_messages[0]['timestamp'],"%d/%m/%y, %H:%M")
print(first_date)
# last date
last_date=datetime.strptime(parsed_messages[-1]['timestamp'],"%d/%m/%y, %H:%M")
print(last_date)
diffrence=last_date - first_date #59 days, 22:14:00
# total days chat:
total_days=diffrence.days+1
print(total_days)
```

```python
person={
    'Rahul':0,
    'Priya':1,
    'Neha':2,
    'Aman':3,
    'Karan':4,
    'Vikas':5
}
#creating matrix 6 x 24:
import numpy as np
heat_map=np.zeros((6,24),dtype=int)
# now retraive parsed message:

for msg in parsed_messages:
  sender=msg['sender']
  dt = datetime.strptime(msg["timestamp"], "%d/%m/%y, %H:%M")
  hour=dt.hour
  # taking row and column from all message:
  row=person[sender]
  column=hour
  heat_map[row][column] +=1
```

##5 World count

```python
## Stoping word this world are comman in every senctence:
stop_words = {
    "i","is","the","a","and","or",
    "to","of","in","on",'for','was',
    'this','how','you','have','about','am',
    'at','my','his','that','he',
    "how","what","who","when","why","which","me",
    "just","so","too","very","only",
    "up","down","out","into","about",'we',
    "are","had","from","been","there","no",
    "with","it","but","one","who","which","be",
    "my","me","our","us","everyone","today"
}
count_word={}
for msg in parsed_messages:
  if msg["message"] in ["<Media omitted>", "This message was deleted"]:
    continue
  text=msg['message']
  text=text.lower()
  # skiping penctuation:
  text=text.replace("!","")
  text=text.replace(",","")
  text=text.replace(".","")
  text=text.replace(":","")
  text=text.replace("?","")
  # split this sentance int single wolds
  words=text.split()
  # now counting words:
  for word in words:
    if word in stop_words:
      continue
    if word in count_word:
      count_word[word] +=1
    else:
      count_word[word] =1

# sorted dictionary
sorted_words = sorted(count_word.items(),key=lambda x: x[1],reverse=True)

print(sorted_words)
```

```python
print("THIS GROUP'S FAVOURITE WORDS\n")

max_count=sorted_words[0][1]
# top 10 most use  words in chat:
for word, count in sorted_words[:10]:
    bar = "█" * int((count / max_count) * 20)
    print(f"{word:<12} {bar:<20} {count}")
```

```python
from datetime import datetime,timedelta

response_time={}
response_count={}

for i in range(len(parsed_messages)-1):
  current = parsed_messages[i]
  next=parsed_messages[i+1]
  # ignoring same sender
  if current["sender"]==next["sender"]:
    continue
  current_time = datetime.strptime(current["timestamp"], "%d/%m/%y, %H:%M")
  next_time = datetime.strptime(next["timestamp"], "%d/%m/%y, %H:%M")
  gap = (next_time - current_time).total_seconds()

  replier = next['sender']

  if replier not in response_time:
    response_time[replier]=0
    response_count[replier]=0
  response_time[replier] +=gap
  response_count[replier] +=1
print("\nRESPONSE PATTERNS\n")
average_response = {}

for person in response_time:
    average = response_time[person] / response_count[person]
    average_response[person] = average
sorted_response = sorted(average_response.items(),key=lambda x: x[1])
fastest=sorted_response[0]
slowest =sorted_response[-1]

fast_minutes=fastest[1]/60
slow_hours=slowest[1]/3600
print(f"fastest replier : {fastest[0]} ({fast_minutes:.1f} minutes)")
print(f"slowest replier : {slowest[0]} ({slow_hours:.1f} hours)")
```

```python
person_days = {}

for msg in parsed_messages:
  person = msg["sender"]
  day = datetime.strptime(msg["timestamp"],"%d/%m/%y, %H:%M").date()
  if person not in person_days:
    person_days[person] = []

  if day not in person_days[person]:
    person_days[person].append(day)


print("LONGEST SILENT STREAKS (consecutive days with zero messages)")

silent = {}

for person, days in person_days.items():
  days.sort()
  longest = 0
  start=None
  end=None

  for i in range(len(days) - 1):
    gap = (days[i + 1] - days[i]).days - 1
    if gap > longest:
      longest=gap
      start=days[i] + timedelta(days=1)
      end=days[i + 1] - timedelta(days=1)


  silent[person]=(longest,start,end)

sorted_silent = sorted(silent.items(),key=lambda x: x[1][0],reverse=True)
for i,(person,data) in enumerate(sorted_silent):
  streak=data[0]
  start=data[1]
  end=data[2]
  if i==0 and streak>0:
    print(f"{person} : {streak} days ({start.strftime('%d %b')} to {end.strftime('%d %b')})")
  elif streak==0:
    print(f"{person} : 0 days (never went silent)")
  else:
    print(f"{person} : {streak} days")
```

```python
from datetime import datetime

people = ["Rahul","Priya","Neha","Aman","Karan","Vikas"]

message_count = {}
night_count = {}
word_count = {}
care_count = {}
caps_count = {}
spam_runs = {}

care_words = [
    "care","sleep","eat","water","drink",
    "safe","home","study","rest","help"
]

for person in people:
  message_count[person] = 0
  night_count[person] = 0
  word_count[person] = 0
  care_count[person] = 0
  caps_count[person] = 0
  spam_runs[person] = []

for msg in parsed_messages:
  person = msg["sender"]
  text = msg["message"]
  message_count[person] += 1

  # Storyteller
  word_count[person] += len(text.split())

  # Night Owl
  hour = datetime.strptime(msg["timestamp"],"%d/%m/%y, %H:%M").hour

  if hour >= 23 or hour <= 4:
    night_count[person] += 1

  # Group Mom
  lower = text.lower()

  for word in care_words:
    if word in lower:
      care_count[person] += 1

  # Drama Queen
  if text.isupper() and len(text) > 3:
    caps_count[person] += 1

# Spammer (average consecutive messages)

current = parsed_messages[0]["sender"]
count = 1

for i in range(1, len(parsed_messages)):
    if parsed_messages[i]["sender"] == current:
      count += 1
    else:
      spam_runs[current].append(count)
      current = parsed_messages[i]["sender"]
      count = 1

spam_runs[current].append(count)
spam_avg = {}

for person in people:
  if len(spam_runs[person]) == 0:
    spam_avg[person] = 0

  else:
    spam_avg[person] = (sum(spam_runs[person])/ len(spam_runs[person]))

# Calculate percentages

night_percent = {}
story_avg = {}
care_percent = {}
caps_percent = {}

for person in people:
  total = message_count[person]
  if total==0:
    continue

  night_percent[person] = night_count[person] * 100 / total
  story_avg[person] = word_count[person] / total
  care_percent[person] = care_count[person] * 100 / total
  caps_percent[person] = caps_count[person] * 100 / total

# Silent Percentage
total_days = (last_date - first_date).days + 1
active_days = {}
for person in people:
  active_days[person] = set()
for msg in parsed_messages:
  person = msg["sender"]
  day = datetime.strptime(msg["timestamp"],"%d/%m/%y, %H:%M").date()
  active_days[person].add(day)

ghost_percent = {}

for person in people:
  silent_days = total_days - len(active_days[person])
  ghost_percent[person] = (silent_days * 100 / total_days)

# Winners
spammer = max(spam_avg.items(), key=lambda x: x[1])
group_mom = max(care_percent.items(), key=lambda x: x[1])
night_owl = max(night_percent.items(), key=lambda x: x[1])
storyteller = max(story_avg.items(), key=lambda x: x[1])
drama = max(caps_percent.items(), key=lambda x: x[1])
ghost = max(ghost_percent.items(), key=lambda x: x[1])
print("\nPERSONALITY ARCHETYPES\n")
print(f"{spammer[0]} -> THE SPAMMER (avg {spammer[1]:.1f} msgs in a row)")
print(f"{group_mom[0]} -> THE GROUP MOM ({group_mom[1]:.1f}% caring messages)")
print(f"{night_owl[0]} -> THE NIGHT OWL ({night_owl[1]:.1f}% msgs after 11 PM)")
print(f"{storyteller[0]} -> THE STORYTELLER (avg {storyteller[1]:.1f} words/msg)")
print(f"{drama[0]} -> THE DRAMA QUEEN ({drama[1]:.1f}% ALL-CAPS msgs)")
print(f"{ghost[0]} -> THE GHOST ({ghost[1]:.1f}% silent days)")
```
